```python
import yfinance as yf
import pandas as pd
import numpy as np

# Incolla qui i ticker esatti del tuo file CSV da 500k
tickers = [
    "VWCE.DE", "CSSPX.MI", "EIMI.MI", "XDWT.DE", "SEGA.MI", 
    "VAGF.DE", "XGLE.MI", "GLD", "ISP.MI", "ENEL.MI", 
    "RACE.MI", "BTC-USD", "ETH-USD"
]

print("Scaricamento dati grezzi...")
raw_data = yf.download(tickers, start="2021-01-01", end="2026-01-01")['Close']

# Mostriamo le prime righe: vedrai molti NaN
display(raw_data.head())
```

```python
print("--- ANALISI DEI DATI MANCANTI ---")
# Contiamo quanti NaN ci sono per ogni colonna
nan_counts = raw_data.isna().sum()
total_rows = len(raw_data)

for ticker, missing in nan_counts.items():
    missing_pct = (missing / total_rows) * 100
    print(f"{ticker}: {missing} giorni mancanti ({missing_pct:.1f}%)")
    
    if missing == total_rows:
        print(f"🚨 ALLARME ROSSO: {ticker} non ha restituito alcun dato! Questo causa il NaN globale.")
        
```
